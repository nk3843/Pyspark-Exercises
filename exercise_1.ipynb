```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("People").getOrCreate()
```

```python
df = spark.read.csv("People.csv",header=True, inferSchema=True)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
filtered_df = df.filter(df.age >= 30)
```

```python
selected_df = filtered_df.select(["name", "city"])
```

```python
selected_df.show()
```

```python
from pyspark.sql.functions import col
```

```python
new_df = df.withColumn("age_plus_5", df.age+5)
```

```python
new_df.show()
```

```python
new_df = df.withColumn("age_plus_5", col("age")+5)
```

```python
new_df.show()
```
