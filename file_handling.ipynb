```python
from pyspark.sql import SparkSession
```

```python
from pyspark.sql.functions import 
```

```python
spark = SparkSession.builder.appName("File_Handling").getOrCreate()
```

```python
json_data = [
    {"student": "Alice", "subject": "Math", "score": 85},
    {"student": "Bob", "subject": "Science", "score": 92},
    {"student": "Charlie", "subject": "English", "score": 78},
    {"student": "Alice", "subject": "Science", "score": 88},
    {"student": "Bob", "subject": "Math", "score": 76}
]
```

```python
students_json_df = spark.createDataFrame(json_data)
```

```python
students_json_df.show()
```

```python
students_json_df.write.json("students.json", mode= "overwrite")
```

```python
students_json_read_df = spark.read.json("students.json")
```

```python
students_json_read_df.printSchema()
```

```python
students_json_df.write.parquet("students.parquet")
```

```python
students_parquet_read_df= spark.read.parquet("students.parquet")
```

```python
students_parquet_read_df.show()
```

```python
students_parquet_read_df.printSchema()
```
