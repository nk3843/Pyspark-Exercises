```python
# Write PySpark code that:
# Reads the customers.csv file into a DataFrame.
# Shows the DataFrame.
# Removes rows that contain any null values.
# Fills the null values in the "age" column with the average age.
# Fills the null values in the "city" column with the string "Unknown".
# Shows the resulting DataFrames after each operation.
```

```python
from pyspark.sql import SparkSession
```

```python
from pyspark.sql.functions import col, avg
```

```python
spark = SparkSession.builder \
    .master("local[*]") \
    .appName("customerApp") \
    .config("spark.ui.port", "4050") \
    .getOrCreate()
```

```python
df = spark.read.csv("customers.csv", header= True, inferSchema=True)
```

```python
df.show()
```

```python
df_filtered = df.dropna()
```

```python
df_filtered.show()
```

```python
average_age = df.select(avg("age")).collect()[0][0]
```

```python
average_age
```

```python
df_filled = df.na.fill(average_age, subset = ["age"])
```

```python
df_filled.show()
```

```python
df_filled = df_filled.na.fill("unknown", subset = ["city"])
```

```python
df_filled.show()
```
