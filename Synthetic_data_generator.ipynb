```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

num_users = 1000
num_events = 10000

user_ids = np.random.randint(1, num_users + 1, num_events)
event_times = [datetime.now() - timedelta(minutes=random.randint(1, 1440)) for _ in range(num_events)]  # Events within 24 hours
event_types = np.random.choice(["page_view", "add_to_cart", "purchase"], num_events)
page_ids = np.random.randint(1, 101, num_events)
product_ids = np.random.randint(1, 51, num_events)

data = {
    "user_id": user_ids,
    "event_time": event_times,
    "event_type": event_types,
    "page_id": page_ids,
    "product_id": product_ids,
}

df = pd.DataFrame(data)
df.to_csv("ecommerce_events.csv", index=False)
```
